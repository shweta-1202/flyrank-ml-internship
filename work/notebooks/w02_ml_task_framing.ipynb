## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is content refresh prioritization. I would treat this as a ranking problem because I want to rank webpages from higher priority to lower priority. This can help the content team decide which pages should be reviewed first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
import os
import subprocess

if not os.path.exists("flyrank-ml-internship-starter"):
    subprocess.run([
        "git", "clone",
        "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
    ], check=True)

os.chdir("flyrank-ml-internship-starter")

print("Data folder found:", os.path.exists("data/raw"))
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining"] = (
    df["trend_direction"].str.lower() == "down"
).astype(int)

print(df["is_declining"].value_counts())
```

My target is whether a webpage is declining. I will create an is_declining target from the trend_direction column. A value of 1 means the page is declining and 0 means it is not declining. This is an observed outcome from the starter dataset.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

My main success metric is Precision@50. It tells me how many of the top 50 ranked pages are actually declining. A higher Precision@50 means the ranking is better at putting useful pages near the top of the review list.

```python
print("Total pages:", len(df))
print("Declining pages:", df["is_declining"].sum())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule uses a few manually chosen conditions, such as page age and impressions. However, webpages can have different combinations of age, impressions, CTR, position, and word count. ML can use several signals together and learn patterns from the available data. This can make the ranking more useful than relying on one simple rule, although the model still needs proper validation.

```python
features = [
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "word_count"
]

df[features].describe()
```
